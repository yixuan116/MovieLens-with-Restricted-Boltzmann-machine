# 34 -- Section 01e Step 4: hidden state

Computes the four joint-state probabilities P(s+, s-) for the Rumble in the
Bronx x h1 worked example, 01e scenario only, reusing q+ and q- saved by
notebook 33. Does not touch the HTML.

## Part 0 -- Setup: scan notebooks/outputs, pick this notebook's number

```python
from pathlib import Path
import json
import re

root = Path.cwd().resolve()
if root.name == "notebooks":
    root = root.parent
notebooks_dir = root / "notebooks"
data_dir = root / "data"
out_dir = root / "outputs"
out_dir.mkdir(parents=True, exist_ok=True)

nb_paths = sorted(p for p in notebooks_dir.glob("*.ipynb") if "step4_hidden_state" not in p.stem)
num_re = re.compile(r"^(\d+)")
numbered = sorted(
    (int(m.group(1)), p) for p in nb_paths if (m := num_re.match(p.stem))
)
highest_num, highest_path = numbered[-1]
this_num = highest_num + 1
this_name = f"{this_num}_step4_hidden_state.ipynb"
print(f"highest existing notebook: {highest_num:>3d}  ({highest_path.name})")
print(f"this notebook is:          {this_num:>3d}  ({this_name})")
```

## Part 1 -- Reuse notebook 33's q+ and q- (01e scenario only)

```python
results33_path = out_dir / "33_step3_hidden_probability_results.json"
assert results33_path.exists(), f"Missing {results33_path}"
results33 = json.loads(results33_path.read_text())

q_plus = results33["q_plus"]["01e"]
q_minus = results33["q_minus"]["01e"]
print(f"q+ (01e) = {q_plus:.6f}")
print(f"q- (01e) = {q_minus:.6f}")
```

## Part 2 -- Joint probability of each of the four states

`s+` and `s-` are drawn independently, so the joint probability factorizes.

```python
P_00 = (1 - q_plus) * (1 - q_minus)
P_10 = q_plus * (1 - q_minus)
P_01 = (1 - q_plus) * q_minus
P_11 = q_plus * q_minus

total = P_00 + P_10 + P_01 + P_11
print(f"P(0,0) = {P_00:.6f}")
print(f"P(1,0) = {P_10:.6f}")
print(f"P(0,1) = {P_01:.6f}")
print(f"P(1,1) = {P_11:.6f}")
print(f"sum    = {total:.6f}")
assert abs(total - 1.0) < 1e-9, "the four joint probabilities must sum to 1"
```

## Part 3 -- Save for Step 4 of section 01e

```python
results = {
    "notebook": this_name,
    "reused_from": "33_step3_hidden_probability.ipynb / outputs/33_step3_hidden_probability_results.json",
    "movie": results33["movie"],
    "user": results33["user"],
    "hidden_unit": results33["hidden_unit"],
    "scenario": "01e",
    "q_plus": q_plus,
    "q_minus": q_minus,
    "P_00": P_00,
    "P_10": P_10,
    "P_01": P_01,
    "P_11": P_11,
    "sum": total,
}
results_path = out_dir / "34_step4_hidden_state_results.json"
with results_path.open("w") as f:
    json.dump(results, f, indent=2)
print(f"saved {results_path}")
print(json.dumps(results, indent=2))
```
